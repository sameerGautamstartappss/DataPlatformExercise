```python
# Read data from a source: Extract
import pandas as pd

# Simulated raw data

data = {
    "id": [1,2,3],
    "name": ["abc", "jkl", "uvw"],
    "age": [25, 30, 35]
}

df = pd.DataFrame(data)
df
```

```python
# Transform: Clean/Modify the data
# Example: add a new column
df["age_category"] = df["age"].apply(lambda x : "Young" if x < 30 else "Adult")
df
```

```python
import mysql.connector

conn = mysql.connector.connect(
    host = "mysql_container1",
    user = "projectuser",
    password = "projectpass",
    database = "projectdb"
)

cursor = conn.cursor()

cursor.execute("""CREATE TABLE IF NOT EXISTS users (
id INT PRIMARY KEY,
name VARCHAR(100),
age INT,
age_category VARCHAR(30)
);""")

# Insert data row by row
for _, row in df.iterrows():
    cursor.execute("REPLACE INTO users (id, name, age, age_category) VALUES (%s, %s, %s, %s)", (row['id'], row['name'], row['age'], row['age_category']))

conn.commit()
cursor.close()
conn.close()
print("Data loaded into MySQL.")
```
